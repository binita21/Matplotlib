```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

### Basic Graph

```python
x=[1,2,3]
y=[1,4,9]
# plt.plot(x,y,label='2X',color='red',linewidth=3,marker='.',linestyle='--',markersize=10,markeredgecolor='Blue')
#Use shorthand notation
#fmt='[color][marker][line]'

plt.figure(figsize=(5,3),dpi=300)

plt.plot(x,y,'r^--',label='2x')
x2=np.arange(0,4.5,0.5)
plt.plot(x2[:6],x2[:6]**2,'b',label='x^2')
plt.plot(x2[5:],x2[5:]**2,'r--')
plt.title("Line Graph!",fontdict={'fontname':'Comic Sans MS','fontsize':20})

plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.xticks([0,1,2,3,4])
plt.yticks([0,2,4,6,8,10])
plt.show() 
```

```python
### Bar chart

labels=['A','B','C']
values=[1,4,2]

bars=plt.bar(labels,values)
patterns = ['/','o','*']
for bar in bars:
    bar.set_hatch(patterns.pop(0))
# bars[0].set_hatch('/')
# bars[1].set_hatch('-')
# bars[2].set_hatch('o')
plt.figure(figsize=(6,4))
plt.show()
```

#### FIFA DATA

```python
fifa = pd.read_csv('data.csv')
fifa
```

### HISTOGRAM

```python
bins=[40,50,60,70,80,90,100]
plt.hist(fifa.Overall,bins=bins,color='#1f004f')
plt.xticks(bins)
plt.ylabel('Number of players')
plt.xlabel('Skill level')
plt.title('Distribution of player skills in FIFA 2018')
plt.show()
```

```python
left=fifa.loc[fifa[ 'Preferred Foot']=='Left'].count()[0]
right=fifa.loc[fifa['Preferred Foot']=='Right'].count()[0]
labels=['Left','Right']
colors=['#abcdff','#ffaabb']
plt.title("Foot Preference of Football Players",fontdict={
'fontfamily':'Times New Roman','fontweight':'bold'})
plt.pie([left,right],labels=labels,colors=colors,autopct='%.2f %%')
plt.show()
```

```python
fifa.Weight = [int(x.strip('lbs')) if type(x)==str else x for x in fifa.Weight]

#Set the default color using ggplot
plt.style.use('ggplot')

# Find the no.of people with different weight categorying light,light_medium, medium, heavy,medium_heavy
fifa.Weight[0]
light_players=fifa.loc[fifa.Weight<125].count()[0]
light_medium=fifa.loc[(fifa.Weight<=125) & (fifa.Weight<150)].count()[0]
medium=fifa.loc[(fifa.Weight>=150) & (fifa.Weight<175)].count()[0]
medium_heavy=fifa.loc[(fifa.Weight>=175)& (fifa.Weight<200)].count()[0]
heavy=fifa.loc[fifa.Weight>=200].count()[0]
print(light_players,light_medium,medium,medium_heavy,heavy)

#Set the labels
labels=['under 125','125-149','150-174','175-199','over 200']

#Explode helps to split up the pie 
explode=(.4,.7,0,0,.6)

plt.title('Players Weight Category')

#Plotting a pie chart 
#pctdistance -> distance of percentage value such as 12.61% from center of pie
#autopct = helps to format the pie in percentage format , here it formatted in 2 decimal values
plt.pie([light_players,light_medium,medium,medium_heavy,heavy],labels=labels,autopct='%.2f %%',pctdistance=0.8,explode=explode)
plt.show()
```

### BOX Plot

```python
barcelona=fifa.loc[fifa.Club=='FC Barcelona']['Overall']
madrid=fifa.loc[fifa.Club=='Real Madrid']['Overall']
revs=fifa.loc[fifa.Club=='New England Revolution']['Overall']
```

```python
plt.style.use('default')
labels=['FC Barcelona','Real Madrid','New England Revolution']
boxes=plt.boxplot([barcelona,madrid,revs],labels=labels,patch_artist=True,medianprops={'linewidth':2})

for box in boxes['boxes']:
    #Set edge color and width
    box.set(color='#4234ff',linewidth=2)

    #Change fill color
    box.set(facecolor='#e0e0e0')
plt.title('Professional Soccer Team COmparison')
plt.ylabel('FIFA Overall Rating')
plt.show()
```
