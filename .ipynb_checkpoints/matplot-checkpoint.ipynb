### MATPLOTLIB

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
%matplotlib inline
```

```python
x=np.arange(-10,11)
x
```

```python
plt.figure(figsize=(12,6))
plt.title("My First Plot")
plt.plot(x,x**2)
plt.plot(x,-1*(x**2))
```

#### Various plots in matplotlib

#### Line plot

```python
x=[0,1,2,3,4]
y=[0,1,4,9,16]
plt.plot(x,y)
```

#### Pie plot

```python
labels=['Python','Numpy','Pandas','C#','Javascript']
sizes = [40, 20, 20, 10,10]
plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.show()
```

#### Bar chart plot

```python
plt.bar(labels,sizes)
```

### Scatter plot

```python
marks1=[20,30,40,10,3]
marks2=[20,20,30,35,40]
plt.scatter(marks1,marks2)
```

#### Histogram plot

```python
data=np.random.randn(100)
plt.hist(data,bins=30)
```

```python
x=[0,2,4,6,8]
y=[0,4,16,36,64]
fig,ax=plt.subplots()
ax.plot(x,y,marker='o',label="Data Points")
ax.set_title("Basic components of matplotlib figure")
ax.set_xlabel("X-Axis")
ax.set_label("Y-Axis")
ax.legend()
plt.show()
```

#### Box Plot

```python
data=[[10,12,14,15,18,20,22],[8,9,11,13,17,19,21],[14,16,18,20,23,25,27]]
plt.boxplot(data,notch=False
            ,vert=False)
```

#### Heatmap plot

```python
x=np.random.seed(0)
data=np.random.rand(10,10)
plt.imshow(data,cmap='viridis',interpolation='nearest')
plt.colorbar()
```

```python
plt.plot([-10,0,10],[-50,-50,-50])
```

```python
plt.figure(figsize=(12,6))
plt.title("Plotting my first chart")
x=np.arange(-10,11)
plt.subplot(1,2,1) #rows,columns,panel selected
plt.plot(x,x ** 2)
plt.plot([0,0,0],[-10,0,100])
plt.legend(['X^2','Vertical Line'])
plt.xlabel('X')
plt.ylabel('X squared')


plt.subplot(1,2,2)
plt.plot(x,-1*(x**2))
plt.plot([-10,0,10],[-50,-50,-50])
plt.legend(['-X^2','Horizontal Line'])
plt.xlabel('X')
plt.ylabel('X squared')
```

### Reading csv and text files


```python
with open('gas_prices.csv','r') as fp:
    print(fp)
```

```python
with open('gas_prices.csv','r') as fp:
    for index,line in enumerate(fp.readlines()):
        if(index<10):
            print(index,line)
```

```python
import csv
```

```python
with open('btc-market-price.csv','r') as fp:
    reader=csv.reader(fp)
    for index,(timestamp,price) in enumerate(reader):
        if(index<10):
            print(f"{timestamp}:${price}")
```

```python
with open('exam_review.csv','r') as fp:
    reader=csv.reader(fp,delimiter='>')
    next(reader)
    for index,values in enumerate(reader):
        print(index,values)
```

#### Using Pandas read_sql method

```python
conn =sqlite3.connect('chinook.db')
```

```python
df=pd.read_sql('SELECT * FROM employees;',conn,index_col='EmployeeId',parse_dates=['BirthDate','HireDate'])
df.head()
```

### using pandas read_sql_query method

```python
conn = sqlite3.connect('chinook.db')
df=pd.read_sql_query('SELECT * FROM employees LIMIT 5;',conn)
df.head()
```

```python
df=pd.read_sql_query('SELECT * FROM employees;',conn,
                    index_col='EmployeeId',
                    parse_dates=['BirthDate','HireDate'])
df.head()
```

```python
x=np.linspace(0,2*np.pi,5)
y=np.sin(x**2)
```

```python
fig,(ax1,ax2)=plt.subplots(1,2)
fig.set_facecolor("red")
ax1.set_facecolor("blue")
ax1.plot(x,y)
ax2.set_facecolor("yellow")
```

```python
fig,axs = plt.subplots(2,2)
axs[0,0].plot(x,x*2,color="red")
axs[0,0].set_title("x*2")
axs[0,1].bar(x,x**2,color="blue")
axs[0,1].set_title("x**2")
axs[1,0].hist(x,x**3,color="green")
axs[1,0].set_title("x**3")
axs[1,1].plot(x,x**4,color="yellow")
axs[1,1].set_title("x**4")
```
